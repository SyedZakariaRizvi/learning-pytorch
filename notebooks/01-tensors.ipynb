### Tensors

A **tensor** is basically a data structure → a generalization of multi-dimensional arrays, used for mathematical operations and computational efficiency.

**Dimension** → how many directions a tensor spans (i.e., its number of axes)

| Dim | Example |
|---|---|
| 0-D | Scalar |
| 1-D | Vectors, word embeddings |
| 2-D | Grayscale image |
| 3-D | RGB image |
| 4-D | Batches of RGB images |
| 5-D | Batch of video clips (see below) |

**5-D example** → batch of 10 video clips, each with 16 frames of size 64×64 and 3 channels
- shape: `[10, 16, 64, 64, 3]`
- read as: `[batch, frames, height, width, channels]`

#### Why are tensors so useful?
1. **Very efficient** at operations like addition, multiplication and dot product → all necessary for NN operations
2. **Real-world data** can be represented easily using tensors
3. Can run on **GPUs** and **parallelize** them too

#### Where are tensors used in DL?
1. **Data storage** → training / testing data
2. **Storing weights and biases**
3. **Matrix operations** → multiplication, dot product, broadcasting
4. **Forward pass and backward pass** (gradients)

## Compying a tensor

```python
a = torch.rand(2, 3)
```

```python
b = a  # not a copy, rather a reference pointing to a

a[0][0] = 0
print(a)
print(b)
```

```python
# both are pointing to the same memory location
print(id(a))
print(id(b))
```

```python
b = a.clone()

a.exp_()
print(a)
print(b)
```

```python
print(id(a))
print(id(b))
```

## Tensor operations on GPU

```python
if torch.cuda.is_available():
    print("GPU is available")
    print(f"Using GPU: {torch.cuda.get_device_name(0)}") 
else:
    print("GPU is not available")
```

```python
if torch.backends.mps.is_available() and torch.backends.mps.is_built():
    print("MPS is available")
else:
    print("MPS is not available")
```

```python
device = torch.device("mps")
```

```python
# creating a new tensor on MPS

torch.rand((2, 3), device=device)
```

```python
# more existing tensor to MPS

a = torch.rand(2, 3)
b = a.to(device)

print(a)
print(b)
```

```python
# time comparison

import time

a = torch.rand((5000, 4000))
b = torch.rand((5000, 4000))

start_time = time.time()
res_cpu = torch.matmul(a, b.T) 
cpu_time = time.time() - start_time

print(f"Time on CPU: {cpu_time}")

c = a.to(device)
d = b.to(device)

torch.mps.synchronize()
start_time = time.time()

res_cpu = torch.matmul(c, d.T) 

torch.mps.synchronize()
mps_time = time.time() - start_time

print(f"Time on MPS: {mps_time}")
```
